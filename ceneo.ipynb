### Parsowanie strony z opiniami o produkcie

```python
from bs4 import BeautifulSoup

page_dom = BeautifulSoup(r.text, 'html.parser')
```

```python
opinions = page_dom.select("div.js_product-review:not(.user-post--highlight)")
print(len(opinions))
type(opinions)
```

```python
opinion = page_dom.select_one("div.js_product-review")
```

```python
opinion = opinions.pop(0)
opinion
```

### Analiza struktury pojedynczej opinii

| składowa | nazwa | selektor |
|---|---|---|
| opinia | opinion | `div.user-post.user-post__card` |
| identyfikator | opinion_id | `div.user-post.user-post__card[data-entry-id]` → atrybut `data-entry-id` |
| autor | author | `span.user-post__author-name` |
| treść | content | `div.user-post__text` |
| ocena | score | `span.user-post__score-count` |
| rekomendacja | recommendation | `em.recommended` |
| lista zalet | pros | `div.review-feature__item.review-feature__item--positive` |
| lista wad | cons | `div.review-feature__item.review-feature__item--negative` |
| dla ilu przydatna | useful | `button.vote-yes span[id^="votes-yes-"]` |
| dla ilu nieprzydatna | unuseful | `button.vote-no span[id^="votes-no-"]` |
| data zamieszczenia | publish_date | `span.user-post__published > time:nth-child(1)[datetime]`|
| data zakupu (czas użytkowania) | purchase_date | `span.user-post__published > time:nth-child(2)[datetime]` |

```python
all_opinions = []


for opinion in opinions:
    
    opinion_data = {}

    # identyfikator opinii
    opinion_data["id"] = opinion.get("data-entry-id")

    opinion_data['author'] = opinion.select_one("span.user-post__author-name").text.strip()

    opinion_data['content'] = opinion.select_one("div.user-post__text").get_text(strip=True)

    
    # Ocena
    opinion_data['score']= opinion.select_one(".user-post__score-count").get_text(strip=True)
    opinion_data['score'] = float(opinion_data['score'].split("/")[0].replace(",","."))

    try:
        opinion_data["recommendation"] = opinion.select_one(".user-post__author-recomendation > em").get_text(strip=True)
    except AttributeError:
        opinion_data['recommendation'] = None

    opinion_data["pros"] = [p.get_text(strip=True) for p in opinion.select("div.review-feature__item--positive")]

    opinion_data["cons"] = [c.get_text(strip=True) for c in opinion.select("div.review-feature__item--negative")]

    opinion_data["useful"] = int(opinion.select_one("button.vote-yes > span").get_text(strip=True)) 
    
    opinion_data["unuseful"] = int(opinion.select_one("button.vote-no > span").get_text(strip=True))


    opinion_data["published_date"] = opinion.select_one("span.user-post__published > time:nth-child(1)").get('datetime')

    try: 
        opinion_data["purchase_date"] = opinion.select_one("span.user-post__published > time:nth-child(2)").get('datetime')
    except AttributeError:
        opinion_data["purchase_date"] = None

    all_opinions.append(opinion_data)
    print(opinion_data)

```
