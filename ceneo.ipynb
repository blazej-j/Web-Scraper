### Parsowanie strony z opiniami o produkcie

```python
from bs4 import BeautifulSoup

page_dom = BeautifulSoup(r.text, 'html.parser')
```

```python
opinions = page_dom.select("div.js_product-review:not(.user-post--highlight)")
print(len(opinions))
type(opinions)
```

```python
opinion = page_dom.select_one("div.js_product-review")
type(opinion)
```

```python
opinion = opinions.pop(0)
opinion
```

### Analiza struktury pojedynczej opinii

| składowa | nazwa | selektor |
|---|---|---|
| opinia | opinion | `div.user-post.user-post__card` |
| identyfikator | opinion_id | `div.user-post.user-post__card[data-entry-id]` → atrybut `data-entry-id` |
| autor | author | `span.user-post__author-name` |
| treść | content | `div.user-post__text` |
| ocena | score | `span.user-post__score-count` |
| rekomendacja | eecommendation | `em.recommended` |
| lista zalet | pros | `div.review-feature__item.review-feature__item--positive` |
| lista wad | cons | `div.review-feature__item.review-feature__item--negative` |
| dla ilu przydatna | useful | `button.vote-yes span[id^="votes-yes-"]` |
| dla ilu nieprzydatna | unuseful | `button.vote-no span[id^="votes-no-"]` |
| data zamieszczenia | publish_date | `span.user-post__published > time:nth-child(1)[datetime]`|
| data zakupu (czas użytkowania) | purchase_date | `span.user-post__published > time:nth-child(2)[datetime]` |

```python
result = {
    "id": opinion.get("data-entry-id"),
    "autor": opinion.select_one("span.user-post__author-name").get_text(strip=True),
    "treść": opinion.select_one("div.user-post__text").get_text(strip=True),
    "ocena": opinion.select_one("span.user-post__score-count").get_text(strip=True),
    "rekomendacja": bool(opinion.select_one("em.recommended")),
    "zalety": [el.get_text(strip=True) for el in opinion.select("div.review-feature__item.review-feature__item--positive")],
    "wady": [el.get_text(strip=True) for el in opinion.select("div.review-feature__item.review-feature__item--negative")],
    "przydatna": opinion.select_one("button.vote-yes span[id^='votes-yes-']").get_text(strip=True),
    "nieprzydatna": opinion.select_one("button.vote-no span[id^='votes-no-']").get_text(strip=True),
    #"data_zamieszczenia": opinion.select_one("div.user-post__published time:first-of-type").get("datetime"),
    #"data_zakupu": opinion.select_one("div.user-post__published time:last-of-type").get("datetime"),
}

for key, value in result.items():
    print(f"{key}: {value}")
```
